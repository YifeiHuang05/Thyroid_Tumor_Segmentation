```python
from pathlib import Path
from collections import Counter

root=Path(r"C:\Users\250030817\Desktop\EEDP_EID_Project\Thyroid_Tumor_Segmentation\data\ThyroidXL_yolo")

for split in ["train","val","test"]:
    counts=Counter()
    n_images=0
    n_empty=0
    label_dir=root/"labels"/split
    for p in label_dir.glob("*.txt"):
        n_images+=1
        lines=[x.strip() for x in p.read_text().splitlines() if x.strip()]
        if not lines:
            n_empty+=1
            continue
        for line in lines:
            counts[int(line.split()[0])]+=1
    print(f"\n{split}")
    print(f"images: {n_images}")
    print(f"empty labels: {n_empty}")
    print(f"class 0 non-PTC polygons: {counts[0]}")
    print(f"class 1 PTC polygons: {counts[1]}")
```
